# Box viewer

Browse a recording window by window with the boxes from `biodetect segment` drawn on top,
and play the window or a single box (band-passed to the box's frequency range).

Every folder under `ROOT` that holds a `files.csv` is a segmentation run and appears in the
**Run** menu. The defaults point at the synthetic demo; run `python examples/synthetic_demo.py`
from the repository root first, or point `ROOT` at your own `results/` folder.

Views: **Spectrogram** (dB), **Denoised S′** (after modal-noise subtraction) and **Otsu mask**
(after opening and closing), so you can see at which step a box appeared or vanished.

```python
import sys
from pathlib import Path

# lets the notebook run from a clone without `pip install -e .`
sys.path.insert(0, str(Path.cwd().parent / "src"))
%matplotlib inline
```

```python
from biodetect.viewer import BoxViewer, plot_window

ROOT = Path("../results")
BoxViewer(ROOT, window_s=10.0).show()
```

## A static frame

`plot_window` draws the same frame without widgets, which is handy for a report figure.
`min_cells` hides boxes smaller than the speckle cutoff.

```python
fig = plot_window(ROOT / "demo" / "boxes", "synthetic_01", t0=0.0, window_s=20.0, min_cells=50)
```
