# Cluster gallery

Look at and listen to the LAMDA clusters from `biodetect cluster`. Tile **[0]** of a cluster is
its representative: the member with the highest global adequacy degree (GAD), which is what the
paper's expert labels. The other tiles are a random sample of the cluster, so you can judge
whether the representative speaks for it.

Names typed in the gallery are saved to `cluster_names.csv` in the clustering folder.

The defaults point at the synthetic demo (`python examples/synthetic_demo.py`).

```python
import sys
from pathlib import Path

# lets the notebook run from a clone without `pip install -e .`
sys.path.insert(0, str(Path.cwd().parent / "src"))
%matplotlib inline
```

```python
from biodetect.viewer import ClusterGallery, cluster_summary, plot_composition

CLUS = Path("../results/demo/boxes_min50/lamda/rank")      # or .../lamda for the paper's min-max scaling
cluster_summary(CLUS).head(20)
```

## What the clusters follow

Each row is a cluster, each column a frequency band or a duration bin. A cluster that sits in
one column is homogeneous in that property. On field recordings the clusters mostly follow the
frequency band (see the README), so read this before trusting a cluster as one sound type.

```python
fig = plot_composition(CLUS, max_clusters=40)
```

## Gallery

Pick a cluster, resample its members, play them in three ways (box only, the box's band with
padding, or everything with padding) and give the cluster a name.

```python
ClusterGallery(CLUS, n=12).show()
```
